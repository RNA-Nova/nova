```python
from nova_agent import Agent
from nova_ai import TextContent

agent = Agent(
    initial_state={
        "system_prompt": "你是一个乐于助人的助手。",
    }
)

# 订阅所有事件并简单打印
def on_event(event):
    print(f"[{event.type}]")

agent.subscribe(on_event)

# 发起对话（Jupyter 中顶层 await 可直接运行）
await agent.prompt("你好，请介绍一下自己。")
await agent.wait_for_idle()

# 查看最终消息
last = agent.state.messages[-1]
if last.role == "assistant" and last.content:
    print("\n最终回复:", last.content[0].text)
```
