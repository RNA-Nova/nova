````python
"""
System prompt construction and project context loading
"""

import os
from datetime import datetime
from typing import Optional, List
from dataclasses import dataclass

# 从 pi_agent 导入 AgentTool
from pi_agent import AgentTool


@dataclass
class ContextFile:
    """Represents a context file with path and content."""
    path: str
    content: str


@dataclass
class BuildSystemPromptOptions:
    """Options for building the system prompt."""
    custom_prompt: Optional[str] = None
    """Custom system prompt (replaces default)."""
    
    tools: Optional[List[AgentTool]] = None
    """Tools to include in prompt. Each tool has name and description."""
    
    append_system_prompt: Optional[str] = None
    """Text to append to system prompt."""
    
    cwd: Optional[str] = None
    """Working directory. Default: current working directory"""
    
    context_files: Optional[List[ContextFile]] = None
    """Pre-loaded context files."""


def build_system_prompt(options: Optional[BuildSystemPromptOptions] = None) -> str:
    """
    Build the system prompt with tools, guidelines, and context.
    
    Args:
        options: Configuration options for building the prompt.
        
    Returns:
        The constructed system prompt string.
    """
    if options is None:
        options = BuildSystemPromptOptions()
    
    custom_prompt = options.custom_prompt
    tools = options.tools
    append_system_prompt = options.append_system_prompt
    cwd = options.cwd
    context_files = options.context_files
    
    resolved_cwd = cwd if cwd is not None else os.getcwd()
    
    now = datetime.now()
    date_time = now.strftime("%A, %B %d, %Y, %I:%M:%S %p %Z")
    
    append_section = f"\n\n{append_system_prompt}" if append_system_prompt else ""
    
    context_files = context_files if context_files is not None else []
    
    if custom_prompt:
        prompt = custom_prompt
        
        if append_section:
            prompt += append_section
        
        # Append project context files
        if len(context_files) > 0:
            prompt += "\n\n# Project Context\n\n"
            prompt += "Project-specific instructions and guidelines:\n\n"
            for file in context_files:
                prompt += f"## {file.path}\n\n{file.content}\n\n"
        
        # Add date/time and working directory last
        prompt += f"\nCurrent date and time: {date_time}"
        prompt += f"\nCurrent working directory: {resolved_cwd}"
        
        return prompt
    
    # Build tools list - 没有默认工具，没有提供就是空列表
    resolved_tools = tools if tools is not None else []
    
    if len(resolved_tools) > 0:
        tools_list = "\n".join([f"- {tool.name}: {tool.description}" for tool in resolved_tools])
    else:
        tools_list = "(none)"
    
    # Build guidelines based on which tools are actually available
    guidelines_list: List[str] = []
    
    tool_names = [tool.name for tool in resolved_tools]
    
    has_bash = "bash" in tool_names
    has_edit = "edit" in tool_names
    has_write = "write" in tool_names
    has_grep = "grep" in tool_names
    has_find = "find" in tool_names
    has_ls = "ls" in tool_names
    has_read = "read" in tool_names
    
    # File exploration guidelines
    if has_bash and not has_grep and not has_find and not has_ls:
        guidelines_list.append("Use bash for file operations like ls, rg, find")
    elif has_bash and (has_grep or has_find or has_ls):
        guidelines_list.append("Prefer grep/find/ls tools over bash for file exploration (faster, respects .gitignore)")
    
    # Read before edit guideline
    if has_read and has_edit:
        guidelines_list.append("Use read to examine files before editing. You must use this tool instead of cat or sed.")
    
    # Edit guideline
    if has_edit:
        guidelines_list.append("Use edit for precise changes (old text must match exactly)")
    
    # Write guideline
    if has_write:
        guidelines_list.append("Use write only for new files or complete rewrites")
    
    # Output guideline (only when actually writing or executing)
    if has_edit or has_write:
        guidelines_list.append(
            "When summarizing your actions, output plain text directly - do NOT use cat or bash to display what you did"
        )
    
    # Always include these
    guidelines_list.append("Be concise in your responses")
    guidelines_list.append("Show file paths clearly when working with files")
    
    guidelines = "\n".join([f"- {g}" for g in guidelines_list])
    
    prompt = f"""You are an expert coding assistant operating inside a coding agent harness. You help users by reading files, executing commands, editing code, and writing new files.

Available tools:
{tools_list}

In addition to the tools above, you may have access to other custom tools depending on the project.

Guidelines:
{guidelines}"""
    
    if append_section:
        prompt += append_section
    
    # Append project context files
    if len(context_files) > 0:
        prompt += "\n\n# Project Context\n\n"
        prompt += "Project-specific instructions and guidelines:\n\n"
        for file in context_files:
            prompt += f"## {file.path}\n\n{file.content}\n\n"
    
    # Add date/time and working directory last
    prompt += f"\nCurrent date and time: {date_time}"
    prompt += f"\nCurrent working directory: {resolved_cwd}"
    
    return prompt
````

````python
"""
System Prompt Manager - 支持初次激活流程（简化版）

分为两个独立接口：
1. build_onboarding() - 初次激活，只加载 setup.md，告知保存目录
2. build() - 正常使用，加载除 setup 外的所有章节 + 动态部分
"""

import os
import re
import glob
import shutil
from datetime import datetime
from typing import Optional, List, Dict
from dataclasses import dataclass

from pi_agent import AgentTool


@dataclass
class PromptContext:
    """动态上下文"""
    tools: Optional[List[AgentTool]] = None
    cwd: Optional[str] = None


class SystemPromptManager:
    """
    系统提示词管理器
    
    目录结构：
    prompts/
    ├── setup.md              # 初次激活专用
    ├── user/                 # 用户数据（由智能体在初次激活时写入）
    ├── 01-role/
    ├── 10-behavior/
    └── ...
    """
    
    def __init__(self, prompts_dir: str = "./prompts"):
        self.prompts_dir = prompts_dir
        self.setup_file = os.path.join(prompts_dir, "setup.md")
        self.user_dir = os.path.join(prompts_dir, "user")
    
    def _discover_sections(self) -> List[tuple[str, List[str]]]:
        """发现所有 section 文件夹（排除 setup.md）"""
        if not os.path.exists(self.prompts_dir):
            return []
        
        sections = []
        
        for item in sorted(os.listdir(self.prompts_dir)):
            if item == "setup.md":
                continue
            
            section_path = os.path.join(self.prompts_dir, item)
            if not os.path.isdir(section_path):
                continue
            
            md_files = glob.glob(os.path.join(section_path, "*.md"))
            md_files = sorted(md_files)
            
            if md_files:
                section_name = self._clean_name(item)
                sections.append((section_name, md_files))
        
        return sections
    
    def _clean_name(self, name: str) -> str:
        """清理 section 名称"""
        cleaned = re.sub(r"^\d+-", "", name)
        cleaned = cleaned.replace("-", " ").replace("_", " ")
        return cleaned.title()
    
    def _load_files(self, files: List[str]) -> str:
        """加载多个文件内容"""
        parts = []
        for filepath in files:
            with open(filepath, "r", encoding="utf-8") as f:
                content = f.read().strip()
                if content:
                    parts.append(content)
        return "\n\n".join(parts) if parts else ""
    
    def _render_template(self, content: str, context: PromptContext) -> str:
        """模板变量替换"""
        now = datetime.now()
        
        tools = context.tools or []
        if tools:
            tool_list = "\n".join([f"- {t.name}: {t.description}" for t in tools])
        else:
            tool_list = "(no tools)"
        
        vars = {
            "cwd": context.cwd or os.getcwd(),
            "datetime": now.strftime("%Y-%m-%d %H:%M:%S"),
            "date": now.strftime("%Y-%m-%d"),
            "time": now.strftime("%H:%M:%S"),
            "tool_list": tool_list,
        }
        
        def replacer(match):
            key = match.group(1).strip()
            return str(vars.get(key, match.group(0)))
        
        return re.sub(r"{{(.+?)}}", replacer, content)
    
    def _build_dynamic_section(self, context: PromptContext) -> str:
        """构建动态部分"""
        parts = []
        
        tools = context.tools or []
        if tools:
            lines = ["# Tools", ""]
            for tool in tools:
                lines.append(f"- **{tool.name}**: {tool.description}")
            parts.append("\n".join(lines))
        
        now = datetime.now()
        meta = f"""# Meta

- **Working directory**: {context.cwd or os.getcwd()}
- **Current time**: {now.strftime("%A, %B %d, %Y, %I:%M:%S %p %Z")}"""
        parts.append(meta)
        
        return "\n\n".join(parts)
    
    # 接口1：初次激活
    def build_onboarding(self) -> Optional[str]:
        """
        初次激活：只加载 setup.md，告知保存目录
        """
        if not os.path.exists(self.setup_file):
            return None
        
        with open(self.setup_file, "r", encoding="utf-8") as f:
            setup_content = f.read().strip()
        
        if not setup_content:
            return None
        
        # 只告知保存目录，不指定具体文件
        save_instructions = f"""

---

## Save Instructions

After collecting user information, save it to:

**Directory**: `{self.user_dir}/`

You can create any files you deem necessary in this directory.
Use the `write` tool to save files. If a file exists, use `read` then `edit` to update.
"""
        
        return f"{setup_content}{save_instructions}"
    
    def has_setup(self) -> bool:
        """检查是否存在 setup.md"""
        return os.path.exists(self.setup_file)
    
    def is_first_time(self) -> bool:
        """检查是否是初次使用（有 setup.md 但无 user/ 目录）"""
        return self.has_setup() and not os.path.exists(self.user_dir)
    
    # 接口2：正常使用
    def build(self, context: Optional[PromptContext] = None) -> str:
        """正常使用：加载除 setup 外的所有章节 + 动态部分"""
        if context is None:
            context = PromptContext()
        
        sections = self._discover_sections()
        
        section_contents = []
        for section_name, files in sections:
            content = self._load_files(files)
            if content:
                section_contents.append(f"# {section_name}\n\n{content}")
        
        if section_contents:
            static_content = "\n\n---\n\n".join(section_contents)
            static_content = self._render_template(static_content, context)
        else:
            static_content = ""
        
        dynamic_content = self._build_dynamic_section(context)
        
        if static_content:
            return f"{static_content}\n\n---\n\n{dynamic_content}"
        else:
            return dynamic_content
    
    def info(self) -> Dict:
        """查看挂载信息"""
        sections = self._discover_sections()
        return {
            "prompts_dir": self.prompts_dir,
            "has_setup": self.has_setup(),
            "is_first_time": self.is_first_time(),
            "sections": [
                {"name": name, "files": [os.path.basename(f) for f in files]}
                for name, files in sections
            ],
        }
    
    def reset(self) -> bool:
        """重置（删除 user/ 目录，下次使用需重新 onboarding）"""
        if os.path.exists(self.user_dir):
            shutil.rmtree(self.user_dir)
            return True
        return False


# 便捷函数
def build_onboarding(prompts_dir: str = "./prompts") -> Optional[str]:
    """便捷函数：构建初次激活提示词"""
    manager = SystemPromptManager(prompts_dir)
    return manager.build_onboarding()


def build_prompt(prompts_dir: str = "./prompts", **context_kwargs) -> str:
    """便捷函数：构建正常使用提示词"""
    manager = SystemPromptManager(prompts_dir)
    context = PromptContext(**context_kwargs)
    return manager.build(context)
````

````python
from datetime import datetime
from typing import Optional, List

@dataclass
class Message():
    id: int
    content: str
    sender: str
    timestamp: datetime = field(default_factory=datetime.now)
    is_read: bool = False
    tags: List[str] = field(default_factory=list)
    
    def mark_as_read(self) -> None:
        self.is_read = True
    
    def add_tag(self, tag: str) -> None:
        if tag not in self.tags:
            self.tags.append(tag)
    
    def summary(self, max_length: int = 50) -> str:
        """返回消息摘要"""
        text = self.content[:max_length]
        if len(self.content) > max_length:
            text += "..."
        return f"[{self.sender}] {text}"
   
class Message(Data):
    def __init__(
        self, 
        id: int, 
        content: str, 
        sender: str,
        timestamp: datetime = None,
        is_read: bool = False,
        tags: List[str] = None
    ):
        self.id = id
        self.content = content
        self.sender = sender
        self.timestamp = timestamp or datetime.now()
        self.is_read = is_read
        self.tags = tags or []
    
    def mark_as_read(self) -> None:
        self.is_read = True
    
    def add_tag(self, tag: str) -> None:
        if tag not in self.tags:
            self.tags.append(tag)
    
    def summary(self, max_length: int = 50) -> str:
        """返回消息摘要"""
        text = self.content[:max_length]
        if len(self.content) > max_length:
            text += "..."
        return f"[{self.sender}] {text}"
    
    def __repr__(self) -> str:
        return (f"Message(id={self.id}, content='{self.content}', "
                f"sender='{self.sender}', timestamp={self.timestamp}, "
                f"is_read={self.is_read}, tags={self.tags})")
    
    def __eq__(self, other) -> bool:
        if not isinstance(other, Message):
            return False
        return (self.id == other.id and 
                self.content == other.content and
                self.sender == other.sender and
                self.timestamp == other.timestamp and
                self.is_read == other.is_read and
                self.tags == other.tags)
    
    def __hash__(self) -> int:
        return hash((self.id, self.content, self.sender))
````

````python
import os
import json
import shutil
import re
from dataclasses import dataclass, field
from pathlib import Path
from typing import Any, Dict, List, Optional

from mashumaro.mixins.json import DataClassJSONMixin
from mashumaro.config import BaseConfig


# ==================== 纯数据定义（无业务逻辑） ====================

@dataclass
class Tool(DataClassJSONMixin):
    """全局工具定义"""
    name: str
    description: str
    parameters: Optional[Dict[str, Any]] = None


@dataclass
class Section(DataClassJSONMixin):
    """系统提示词片段"""
    name: str           # 清理后的名称（如 "01-role.md" → "role"）
    order: int          # 排序序号
    content: str        # markdown 原始内容
    raw_filename: str   # 原始文件名


@dataclass
class AgentConfig(DataClassJSONMixin):
    """智能体配置快照"""
    name: str
    path: Path
    description: Optional[str] = None
    mounts: List[str] = field(default_factory=list)
    sections: List[Section] = field(default_factory=list)
    
    class Config(BaseConfig):
        serialization_strategy = {
            Path: {
                "serialize": lambda x: str(x),
                "deserialize": lambda x: Path(x)
            }
        }


@dataclass
class UserData(DataClassJSONMixin):
    """用户数据挂载"""
    files: Dict[str, str] = field(default_factory=dict)  # 相对路径: 内容


@dataclass
class PromptsBundle(DataClassJSONMixin):
    """完整的 prompts 目录结构"""
    path: Path
    setup_content: Optional[str] = None
    tools: List[Tool] = field(default_factory=list)
    master: Optional[AgentConfig] = None
    subagents: Dict[str, AgentConfig] = field(default_factory=dict)
    user_data: UserData = field(default_factory=UserData)
    
    class Config(BaseConfig):
        serialization_strategy = {
            Path: {
                "serialize": lambda x: str(x),
                "deserialize": lambda x: Path(x)
            }
        }
        omit_none = True


# ==================== 加载逻辑（独立类） ====================

class PromptsLoader:
    """PromptsBundle 的加载器：负责文件系统 → Dataclass 的转换"""
    
    def __init__(self, prompts_dir: str | Path):
        self.prompts_dir = Path(prompts_dir)
    
    def load(self) -> PromptsBundle:
        """加载完整的 PromptsBundle"""
        tools = self._load_tools()
        
        return PromptsBundle(
            path=self.prompts_dir,
            setup_content=self._load_file(self.prompts_dir / "setup.md"),
            tools=tools,
            master=self._load_agent(self.prompts_dir / "master", "master"),
            subagents=self._load_all_subagents(),
            user_data=self._load_user_data()
        )
    
    def _load_file(self, filepath: Path) -> Optional[str]:
        """通用文件加载"""
        if not filepath.exists():
            return None
        try:
            content = filepath.read_text(encoding="utf-8").strip()
            return content if content else None
        except IOError:
            return None
    
    def _load_tools(self) -> List[Tool]:
        """加载 tools.json"""
        tools_file = self.prompts_dir / "tools.json"
        if not tools_file.exists():
            return []
        
        try:
            raw_data = json.loads(tools_file.read_text(encoding="utf-8"))
            if isinstance(raw_data, list):
                return [Tool.from_dict(item) for item in raw_data]
        except (json.JSONDecodeError, IOError):
            pass
        return []
    
    def _load_mounts(self, mounts_file: Path) -> List[str]:
        """解析 mounts.json"""
        if not mounts_file.exists():
            return []
        
        try:
            data = json.loads(mounts_file.read_text(encoding="utf-8"))
            if isinstance(data, list):
                return data
            elif isinstance(data, dict) and "tools" in data:
                return data["tools"]
        except (json.JSONDecodeError, IOError):
            pass
        return []
    
    def _parse_section(self, filepath: Path) -> Optional[Section]:
        """将单个 .md 文件解析为 Section dataclass"""
        if not filepath.exists() or filepath.suffix != ".md":
            return None
        
        content = self._load_file(filepath)
        if not content:
            return None
        
        raw_name = filepath.stem
        # 解析序号前缀：01-role → order=1, name="role"
        match = re.match(r"^(\d+)[-_]", raw_name)
        if match:
            order = int(match.group(1))
            name = raw_name[match.end():]
        else:
            order = 9999
            name = raw_name
        
        return Section(
            name=name.replace("-", " ").replace("_", " "),
            order=order,
            content=content,
            raw_filename=filepath.name
        )
    
    def _load_agent(self, agent_path: Path, name: str) -> Optional[AgentConfig]:
        """加载单个智能体配置"""
        if not agent_path.exists() or not agent_path.is_dir():
            return None
        
        # 加载 sections
        sections = []
        sections_dir = agent_path / "sections"
        if sections_dir.exists():
            for md_file in sorted(sections_dir.glob("*.md")):
                section = self._parse_section(md_file)
                if section:
                    sections.append(section)
        
        return AgentConfig(
            name=name,
            path=agent_path,
            description=self._load_file(agent_path / "description.md"),
            mounts=self._load_mounts(agent_path / "mounts.json"),
            sections=sections
        )
    
    def _load_all_subagents(self) -> Dict[str, AgentConfig]:
        """加载所有子智能体"""
        subagents_dir = self.prompts_dir / "subagents"
        if not subagents_dir.exists():
            return {}
        
        result = {}
        for agent_dir in sorted(subagents_dir.iterdir()):
            if agent_dir.is_dir():
                agent = self._load_agent(agent_dir, agent_dir.name)
                if agent:
                    result[agent.name] = agent
        return result
    
    def _load_user_data(self) -> UserData:
        """递归加载 user/ 目录"""
        user_dir = self.prompts_dir / "user"
        files: Dict[str, str] = {}
        
        if not user_dir.exists():
            return UserData(files=files)
        
        for root, _, filenames in os.walk(user_dir):
            for filename in sorted(filenames):
                if not filename.endswith(".md"):
                    continue
                
                filepath = Path(root) / filename
                rel_path = filepath.relative_to(user_dir).as_posix()
                content = self._load_file(filepath)
                
                if content:
                    files[rel_path] = content
        
        return UserData(files=files)


# ==================== 可选：写入逻辑（独立类） ====================

class PromptsWriter:
    """PromptsBundle 的写入器：负责内存状态 → 文件系统的持久化"""
    
    def __init__(self, bundle: PromptsBundle):
        self.bundle = bundle
    
    def save_user_file(self, relative_path: str, content: str) -> Path:
        """保存用户数据文件"""
        user_dir = self.bundle.path / "user"
        user_dir.mkdir(parents=True, exist_ok=True)
        
        target = user_dir / relative_path
        target.parent.mkdir(parents=True, exist_ok=True)
        target.write_text(content, encoding="utf-8")
        
        # 同步更新 bundle 内存状态
        self.bundle.user_data.files[relative_path] = content
        return target
    
    def reset_user_data(self) -> bool:
        """删除 user/ 目录并重置内存状态"""
        user_dir = self.bundle.path / "user"
        if user_dir.exists():
            shutil.rmtree(user_dir)
            self.bundle.user_data = UserData()
            return True
        return False
    
    def save_bundle_cache(self, filename: str = ".bundle_cache.json") -> Path:
        """将完整 bundle 序列化为 JSON 缓存"""
        cache_file = self.bundle.path / filename
        cache_file.write_text(
            self.bundle.to_json(indent=2), 
            encoding="utf-8"
        )
        return cache_file


# ==================== 使用示例 ====================

def example():
    # 1. 加载（Loader 负责所有文件操作）
    loader = PromptsLoader("./prompts")
    bundle = loader.load()
    
    # 2. Dataclass 是纯数据，可自由序列化传输/存储
    json_snapshot = bundle.to_json()
    
    # 3. 从 JSON 恢复（无需文件系统）
    restored_bundle = PromptsBundle.from_json(json_snapshot)
    
    # 4. 修改后通过 Writer 持久化
    writer = PromptsWriter(bundle)
    writer.save_user_file("profile.md", "# User Profile\n\nName: Alice")
    
    # 5. 访问纯数据（无逻辑）
    if bundle.master:
        print(f"Master sections count: {len(bundle.master.sections)}")
        
    # 6. 直接构造（无需 Loader，用于测试或程序化生成）
    test_bundle = PromptsBundle(
        path=Path("./test"),
        tools=[Tool(name="test", description="test tool")],
        master=AgentConfig(
            name="master",
            path=Path("./test/master"),
            sections=[
                Section(name="role", order=1, content="You are a helper.", raw_filename="01-role.md")
            ]
        )
    )

if __name__ == "__main__":
    example()
````

````python
import asyncio
import tempfile
import os
from pathlib import Path

# 假设导入路径如下（根据你的实际项目结构调整）
from nova_simple_agent.core.resource import (
    DefaultResourceLoader,
    DefaultResourceLoaderOptions,
)


async def basic_example():
    """基础用法示例 - 使用默认配置加载"""
    # 1. 初始化配置选项（使用当前工作目录和默认的 agent 目录）
    options = DefaultResourceLoaderOptions(
        include_defaults=True,  # 包含默认的提示词目录
    )
    
    # 2. 创建加载器实例
    loader = DefaultResourceLoader(options)
    
    # 3. 执行加载
    await loader.reload()
    
    # 4. 获取结果
    result = loader.get_prompts()
    prompts = result["prompts"]
    diagnostics = result["diagnostics"]
    
    print(f"✅ 加载了 {len(prompts)} 个提示词模板")
    for prompt in prompts:
        print(f"  - {prompt.name} ({prompt.source}): {prompt.description[:50]}...")
    
    if diagnostics:
        print(f"⚠️ 发现 {len(diagnostics)} 个诊断信息")
        for diag in diagnostics:
            print(f"  - [{diag['type']}] {diag['message']}")


async def advanced_example():
    """高级用法示例 - 自定义路径和冲突处理"""
    
    # 创建临时目录模拟自定义提示词
    with tempfile.TemporaryDirectory() as tmpdir:
        # 创建一些示例提示词文件
        prompt_dir = Path(tmpdir) / "custom_prompts"
        prompt_dir.mkdir()
        
        # 创建两个同名不同路径的提示词（模拟冲突）
        (prompt_dir / "code_review.md").write_text("""---
name: code_review
description: 代码审查提示词
---
请审查以下代码的质量...
""")
        
        # 创建第二个同名文件（模拟冲突）
        conflict_dir = Path(tmpdir) / "conflict_prompts"
        conflict_dir.mkdir()
        (conflict_dir / "code_review.md").write_text("""---
name: code_review
description: 另一个代码审查提示词
---
请分析代码的性能问题...
""")
        
        # 配置加载器
        options = DefaultResourceLoaderOptions(
            cwd=str(tmpdir),
            include_defaults=False,  # 不加载默认提示词，只加载指定路径
            prompt_paths=[
                str(prompt_dir),
                str(conflict_dir)
            ]
        )
        
        loader = DefaultResourceLoader(options)
        await loader.reload()
        
        result = loader.get_prompts()
        
        print("\n=== 冲突检测示例 ===")
        print(f"最终加载的提示词数量: {len(result['prompts'])}")
        
        # 检查诊断信息中的冲突
        collisions = [d for d in result["diagnostics"] if d["type"] == "collision"]
        for collision in collisions:
            info = collision["collision"]
            print(f"⚠️ 检测到名称冲突: /{info['name']}")
            print(f"  保留: {info['winner_path']}")
            print(f"  忽略: {info['loser_path']}")


async def selective_loading_example():
    """按需加载示例 - 禁用默认提示词，只加载特定路径"""
    
    options = DefaultResourceLoaderOptions(
        no_prompt_templates=True,  # 禁用所有默认提示词
        prompt_paths=["/path/to/my/custom/prompts"]  # 只加载指定目录
    )
    
    loader = DefaultResourceLoader(options)
    await loader.reload()
    
    result = loader.get_prompts()
    print(f"仅加载自定义提示词: {len(result['prompts'])} 个")


def sync_wrapper_example():
    """同步包装示例（如果在同步上下文中使用）"""
    async def load():
        options = DefaultResourceLoaderOptions()
        loader = DefaultResourceLoader(options)
        await loader.reload()
        return loader.get_prompts()
    
    # 在同步代码中运行异步加载
    result = asyncio.run(load())
    return result["prompts"]


async def hot_reload_example():
    """热重载示例 - 动态重新加载"""
    
    options = DefaultResourceLoaderOptions(include_defaults=True)
    loader = DefaultResourceLoader(options)
    
    # 首次加载
    await loader.reload()
    initial_count = len(loader.get_prompts()["prompts"])
    print(f"初始加载: {initial_count} 个提示词")
    
    # ... 假设此时有新文件添加到提示词目录 ...
    
    # 重新加载（不创建新实例）
    await loader.reload()
    new_count = len(loader.get_prompts()["prompts"])
    print(f"重新加载后: {new_count} 个提示词")


async def error_handling_example():
    """错误处理示例"""
    try:
        options = DefaultResourceLoaderOptions(
            prompt_paths=["/不存在的路径/"],
            include_defaults=False
        )
        loader = DefaultResourceLoader(options)
        await loader.reload()
        
        result = loader.get_prompts()
        
        # 检查诊断信息
        if result["diagnostics"]:
            print("发现诊断信息:")
            for diag in result["diagnostics"]:
                print(f"  [{diag['type']}] {diag['message']} @ {diag['path']}")
        
        # 即使没有错误，也可能返回空列表
        if not result["prompts"]:
            print("警告: 未加载到任何提示词模板")
            
    except Exception as e:
        print(f"加载失败: {e}")


# 运行示例
if __name__ == "__main__":
    print("=== 基础示例 ===")
    asyncio.run(basicav_example())
    
    print("\n=== 高级示例 ===")
    asyncio.run(advanced_example())
    
    print("\n=== 热重载示例 ===")
    asyncio.run(hot_reload_example())
````

````python
import os
import argparse
from pathlib import Path

def py_files_to_md(input_path, output_file="output.md"):
    """
    将指定路径下的所有Python文件内容整合到一个Markdown文件中
    
    Args:
        input_path: 输入的路径（文件或目录）
        output_file: 输出的Markdown文件名
    """
    py_files = []
    
    # 获取所有Python文件
    path = Path(input_path)
    if path.is_file():
        if path.suffix == '.ts':
            py_files.append(path)
    elif path.is_dir():
        py_files = list(path.rglob("*.ts"))
    else:
        print(f"错误：路径 '{input_path}' 不存在")
        return
    
    if not py_files:
        print(f"在路径 '{input_path}' 中没有找到Python文件")
        return
    
    # 写入Markdown文件
    with open(output_file, 'w', encoding='utf-8') as md_file:
        for py_file in sorted(py_files):
            # 写入文件路径作为标题
            relative_path = py_file.relative_to(path.parent if path.is_file() else path)
            md_file.write(f"# {relative_path}\n\n")
            
            # 写入代码块
            try:
                with open(py_file, 'r', encoding='utf-8') as f:
                    content = f.read()
                md_file.write("```python\n")
                md_file.write(content)
                if not content.endswith('\n'):
                    md_file.write('\n')
                md_file.write("```\n\n")
            except Exception as e:
                md_file.write(f"*无法读取文件: {e}*\n\n")
    
    print(f"已完成！共处理 {len(py_files)} 个Python文件，输出到 {output_file}")
````
